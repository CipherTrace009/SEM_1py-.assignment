```python
                          ## ASSIGNMENT NO. 1

# Write a python program to create, append and remove etc.
# operation on Dictionary and Tuple.

## Dictionary Operations

list01 =["name", "age", "coure"]
list02 =["Disha","18", "CSE - CSF"]

my_dict = dict(zip(list01,list02))
print("Original Dictionary:", my_dict)

# Add an element
my_dict["city"] = "Bhopal,MP"
print("Addition:", my_dict)

# Update an element
my_dict["age"] = 19
print("Updating:", my_dict)

# Remove an element
my_dict.pop("city")
print("Removing:", my_dict)

# Accessing an element
print("Accessing an element:", my_dict["name"])

# Length of dictionary
X =len(my_dict)
print(X)
```

```python
## Tuple Operations
my_tup = (1,2,3,4,5,6)
print("Original Tuple:",my_tup)

# Access an element
print("First element:", my_tup [0])

# Slicing
print("Sliced Tuple:", my_tup [-5:-2])

# Concatenation
num = (101,969,33)
new_tuple = my_tup  + num
print("After concatenation:", new_tuple)

# Repetition
print("Repeated Tuple:", (1, 2) * 0)

# Count
print("Count of 20:", my_tup .count(2))

# Find index
print("Index of 33:", my_tup .index(3))
```

```python
                          ## ASSIGNMENT NO. 3

# write a python program that accecpts length of sides of triangle
# as inputs and tells whether or not the triangle is
# right angled or not using functions

def R_tri (a,b,c):
    largest = max(a,b,c)

    if a==largest:
        if (b**2 + c**2 ==a**2):
            return True
    if b==largest:
        if (a**2 + c**2 ==b**2):
            return True
    if c==largest:
        if (a**2 + b**2 ==c**2):
            return True
    else:
        return False
print ("enter the three sides of triangle")
a = int(input("enter first side:"))
b = int(input("enter second side:"))
c = int(input("enter third side:"))

if R_tri (a,b,c):
    print ("it's a right angled triangle")

else:
    print ("it's NOT a right angled triangle")
```

```python
## Addition of two matrices using list
matrix1 = [[1, 2], [3, 4]]
matrix2 = [[5, 6], [7, 8]]

result = []
for i in range(2):
    row = []
    for j in range(2):
        row.append(matrix1[i][j] + matrix2[i][j])
    result.append(row)

print("First matrix:", matrix1)
print("Second matrix:", matrix2)
print("Addition of matrices:", result)
```
